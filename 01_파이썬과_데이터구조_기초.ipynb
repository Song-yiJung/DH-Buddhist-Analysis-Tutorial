## 💡인문학자에게 파이썬(Python)이란?

왜 우리는 '파이썬'을 배워야 할까요? 인문학자가 파이썬을 배운다는 것은, 단순히 프로그래밍 기술을 익히는 것이 아닙니다. 이는 **지치지 않고, 실수 없는 유능한 조수**를 한 명 고용하는 것과 같습니다.

네덜란드계 소프트웨어 엔지니어 **귀도 반 로섬(Guido van Rossum)**이 만든 이 언어는 **사람이 읽기 쉽도록** 설계되었습니다. 컴퓨터의 언어라기보다, 영어로 된 수행 지시문에 가깝습니다.

**특징**

파이썬은 **인터프리터 사용**(통역사가 있어서 코드를 바로바로 실행 가능), **프로그래밍 언어**(컴퓨터랑 대화하는 언어), **객체지향 언어**이기에 코드 재사용 용이하고 원도우나 리눅스 어느 환경에서든 사용 가능, **동적 타이핑 언어**로 인간이 입력한 값을 알아서 숫자인가 글자인가 판단 가능하다.

## 💡디지털 인문학에서의 역할

* **전수 조사:** 사람이 평생 읽어도 못 다 읽을 『조선왕조실록』이나 『대장경』 텍스트 전체에서 특정 단어(예: '사리(舍利)')가 언제, 몇 번, 어떤 맥락으로 나오는지 1초 만에 찾아냅니다.
* **단순 반복 업무 해방:** 수천 건의 엑셀 서지 목록에서 '저자'와 '연도'만 뽑아 정리하는 작업을, 파이썬은 커피 한 모금 마실 시간에 끝냅니다.
* **패턴 발견:** 텍스트를 데이터로 변환하여, 시대별 문체 변화나 특정 개념의 전승 과정을 시각적으로 보여줍니다.

이러한 활용이 가능한 도구인 파이썬은 디지털 인문학에서 다음과 같은 **핵심적인 장점**을 가지고 있습니다.

* **쉬운 학습 곡선:** 문법이 직관적이고 간단해서 프로그래밍 경험이 없는 인문학 연구자들도 빠르게 배우고 적용할 수 있습니다.
* **풍부한 라이브러리 생태계:** 텍스트 분석(NLTK, spaCy, KoNLPy), 데이터 처리(Pandas), 시각화(Matplotlib, Seaborn) 등 각 분야에 특화된 수많은 라이브러리(미리 만들어진 기능 묶음)들이 존재하여, 복잡한 기능도 몇 줄의 코드로 구현할 수 있습니다.
* **확장성 및 유연성:** 다른 언어나 시스템과 연동하기 쉽고, 다양한 연구 방법론에 유연하게 적용할 수 있습니다.

## 파이썬 기초: 문법이 아니라 '언어'로 이해하기

(1) 변수(Variable): 데이터에 이름표 붙이기 컴퓨터에게 "이 데이터는 앞으로 이렇게 부르겠다"라고 약속하는 것입니다.

```python
# 일반적인 예시가 아니라, 연구 맥락으로 이해해 봅시다.

# 'book_title'이라는 이름표(변수)에 "삼국유사"라는 책 제목(데이터)을 붙입니다.
book_title = "삼국유사"

# 'author'라는 이름표에 "일연"을 붙입니다.
author = "일연"

# 'published_year'라는 이름표에 숫자 1281을 붙입니다.
published_year = 1281

# 이름표를 부르면, 컴퓨터는 저장된 내용을 보여줍니다.
print(book_title)
print(author)

# 연구자가 할 수 있는 질문: 이 책은 지금으로부터 몇 년 전 책인가?
current_year = 2025
age = current_year - published_year

print(f"문헌 '{book_title}'는 지금으로부터 약 {age}년 전에 편찬되었습니다.")
```

(2) 함수(Function): 마법의 도구 상자 함수는 **자판기**나 **사전 찾기**와 같습니다. 우리가 원리를 몰라도 버튼을 누르면 결과가 나오듯, 이미 만들어진 기능을 가져다 쓰는 것입니다.


*   print(): 모니터에 결과를 출력해라. (가장 많이 쓰는 도구)
*   len(): 텍스트의 길이를 재라. (Length)
*   type(): 이 데이터가 숫자인지 문자인지 알려달라.


```python
text = "나라 말씀이 중국과 달라 문자와 서로 통하지 아니하여"

# 함수 사용해보기
print(text)           # 화면에 출력
print(len(text))      # 글자 수 세기 (공백 포함)

# 결과: 28 (글자 수)
```

### **4. 데이터 구조: 서지 자료를 담는 그릇**

서지학에서 자료를 카드에 정리하듯, 파이썬도 데이터를 담는 그릇이 중요합니다.

(1) 리스트(List): 순서가 있는 서가 책장에 책을 순서대로 꽂아둔 것과 같습니다. 순서가 중요하고, 언제든 책을 빼거나 더 꽂을 수 있습니다.

```python
# 불교 경전 목록을 리스트로 만들어 봅시다.
sutras = ["금강경", "법화경", "화엄경"]

print(f"내 서가에 있는 경전들: {sutras}")

# 첫 번째 경전 꺼내기 (파이썬은 숫자를 0부터 셉니다!)
print(f"가장 먼저 공부할 책: {sutras[0]}")

# 새로운 책 구입 (리스트에 추가하기)
sutras.append("유마경")
print(f"책 추가 후 목록: {sutras}")
```

(2) 딕셔너리(Dictionary): 서지 카드 (메타데이터) 순서보다는 **표제어**(Key)와 **내용**(Value)의 연결이 중요합니다. 도서관의 서지 정보를 다룰 때 가장 강력한 도구입니다.



```python
# 한 권의 책에 대한 상세 정보를 담습니다.
book_info = {
    "서명": "월인천강지곡",
    "저자": "세종",
    "권수": "상, 중, 하",
    "발행년도": 1447
}

# 필요한 정보만 쏙쏙 뽑아낼 수 있습니다.
print(f"이 책의 제목은 무엇인가? : {book_info['서명']}")
print(f"누가 지었는가? : {book_info['저자']}")
```

### 5. 파이썬의 꽃: 반복문과 조건문 (연구 자동화의 핵심)
이 부분이 바로 **파이썬을 배우는 이유**입니다. 사람이 수작업으로 하던 분류와 검토를 자동화합니다.

(1) 반복문 (for): "이 서가에 있는 책을 처음부터 끝까지 다 검토해라"

```python
# 연구 대상 문헌 목록
research_list = ["훈민정음 해례본", "용비어천가", "월인석보", "석보상절"]

print("--- 문헌 조사 시작 ---")

# 'research_list'에 있는 책을 하나씩 꺼내서 'book'이라는 임시 이름으로 부르며 반복합니다.
for book in research_list:
    print(f"현재 검토 중인 문헌: {book}")
    # 여기에 복잡한 분석 코드가 들어가면, 수천 권도 순식간에 분석합니다.

print("--- 조사 완료 ---")
```

### (2) 조건문 (if): "만약 불교 관련 서적이라면 따로 분류해라"

```python
# 섞여 있는 문헌 제목들
titles = ["삼국유사", "논어", "금강경", "맹자", "화엄경"]

print("--- 불교 서적 분류 작업 ---")

for title in titles:
    # 제목에 '경(Sutra)'이나 '유사'가 들어가는지 기계적으로 판단
    if "경" in title or "유사" in title:
        print(f"⭕ [불교/역사] : {title}")
    else:
        print(f"❌ [유교/기타] : {title}")
```
